```python
# Bubble Sort Visualizer for Jupyter Notebook
# Run this entire code in one Jupyter cell.
# Requires: ipywidgets

import random
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# -----------------------------
# State
# -----------------------------
original_values = [6, 3, 8, 2, 7, 1, 5, 4]
values = original_values[:]
j = 0                       # left element of current adjacent pair
end = len(values) - 1       # last unsorted index
pass_number = 1
finished = False
last_swapped = False
swapped_this_pass = False

# -----------------------------
# UI
# -----------------------------
title = widgets.HTML(
    "<h3 style='margin-bottom:4px'>Bubble Sort Visualizer</h3>"
    "<div>Compare adjacent values. If the left value is bigger, swap them.</div>"
)

status = widgets.HTML()
explanation = widgets.HTML()

next_btn = widgets.Button(description="Next Comparison", button_style="primary")
pass_btn = widgets.Button(description="Finish This Pass")
shuffle_btn = widgets.Button(description="Shuffle")
reset_btn = widgets.Button(description="Reset")

output = widgets.Output()

def render():
    global values, j, end, pass_number, finished, last_swapped

    boxes = []

    for idx, value in enumerate(values):
        if finished or idx > end:
            bg = "#dff5df"
            border = "#6aa66a"
            label = "sorted"

        elif idx == j:
            bg = "#fff0b3"
            border = "#d6a600"
            label = "left"

        elif idx == j + 1:
            bg = "#dcecff"
            border = "#4a90e2"
            label = "right"

        else:
            bg = "#f4f4f4"
            border = "#999"
            label = ""

        boxes.append(
            f"""
            <div style="
                width:70px;
                height:70px;
                display:flex;
                flex-direction:column;
                align-items:center;
                justify-content:center;
                border:3px solid {border};
                background:{bg};
                border-radius:10px;
                margin:5px;
                font-family:Arial;
            ">
                <div style="font-size:26px;font-weight:bold">{value}</div>
                <div style="font-size:11px">{label}</div>
            </div>
            """
        )

    if finished:
        status.value = "<b>Finished!</b> The list is sorted."
        explanation.value = (
            "<div style='margin-top:8px'><b>Bubble Sort:</b> "
            "Large values have bubbled to the right.</div>"
        )
    else:
        left = values[j]
        right = values[j + 1]

        status.value = (
            f"Pass <b>{pass_number}</b>. "
            f"Comparing index <b>{j}</b> and <b>{j + 1}</b>: "
            f"<b>{left}</b> and <b>{right}</b>."
        )

        if left > right:
            action = f"{left} &gt; {right} → they will <b>swap</b>."
        else:
            action = f"{left} ≤ {right} → <b>no swap</b>."

        explanation.value = (
            "<div style='margin-top:8px'>"
            + action +
            "</div>"
        )

    with output:
        clear_output(wait=True)
        display(HTML(
            "<div style='display:flex;flex-wrap:wrap'>"
            + "".join(boxes)
            + "</div>"
        ))

def reset_state(new_values=None):
    global values, j, end, pass_number, finished
    global last_swapped, swapped_this_pass

    if new_values is not None:
        values = new_values[:]

    j = 0
    end = len(values) - 1
    pass_number = 1
    finished = False
    last_swapped = False
    swapped_this_pass = False
    render()

def next_comparison(_=None):
    global values, j, end, pass_number, finished
    global last_swapped, swapped_this_pass

    if finished:
        return

    # Compare adjacent pair and swap immediately if needed
    if values[j] > values[j + 1]:
        values[j], values[j + 1] = values[j + 1], values[j]
        last_swapped = True
        swapped_this_pass = True
    else:
        last_swapped = False

    j += 1

    # End of one pass
    if j >= end:
        # The value at 'end' is now in final position
        if not swapped_this_pass:
            finished = True
            render()
            return

        end -= 1
        pass_number += 1
        j = 0
        swapped_this_pass = False

        if end <= 0:
            finished = True

    render()

def finish_pass(_=None):
    global values, j, end, pass_number, finished
    global swapped_this_pass, last_swapped

    if finished:
        return

    while j < end:
        if values[j] > values[j + 1]:
            values[j], values[j + 1] = values[j + 1], values[j]
            swapped_this_pass = True
        j += 1

    if not swapped_this_pass:
        finished = True
        render()
        return

    end -= 1
    pass_number += 1
    j = 0
    swapped_this_pass = False
    last_swapped = False

    if end <= 0:
        finished = True

    render()

def shuffle_values(_=None):
    nums = list(range(1, 9))
    random.shuffle(nums)
    reset_state(nums)

def reset_values(_=None):
    reset_state(original_values)

next_btn.on_click(next_comparison)
pass_btn.on_click(finish_pass)
shuffle_btn.on_click(shuffle_values)
reset_btn.on_click(reset_values)

controls = widgets.HBox([next_btn, pass_btn, shuffle_btn, reset_btn])

display(title, status, controls, explanation, output)
render()
```

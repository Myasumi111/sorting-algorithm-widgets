```python
import random
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output


def create_bubble_sort_visualizer():

    # -----------------------------
    # State
    # -----------------------------
    original_values = [6, 3, 8, 2, 7, 1, 5, 4]
    values = original_values[:]

    j = 0
    end = len(values) - 1
    pass_number = 1
    finished = False
    last_swapped = False
    swapped_this_pass = False

    # -----------------------------
    # UI
    # -----------------------------
    title = widgets.HTML(
        "<h3 style='margin-bottom:4px'>Bubble Sort Visualizer</h3>"
        "<div>Compare adjacent values. If the left value is bigger, swap them.</div>"
    )

    status = widgets.HTML()
    explanation = widgets.HTML()

    next_btn = widgets.Button(
        description="Next Comparison",
        button_style="primary"
    )

    pass_btn = widgets.Button(
        description="Finish This Pass"
    )

    shuffle_btn = widgets.Button(
        description="Shuffle"
    )

    reset_btn = widgets.Button(
        description="Reset"
    )

    output = widgets.Output()

    # -----------------------------
    # Render
    # -----------------------------
    def render():

        boxes = []

        for idx, value in enumerate(values):

            if finished or idx > end:
                bg = "#dff5df"
                border = "#6aa66a"
                label = "sorted"

            elif idx == j:
                bg = "#fff0b3"
                border = "#d6a600"
                label = "left"

            elif idx == j + 1:
                bg = "#dcecff"
                border = "#4a90e2"
                label = "right"

            else:
                bg = "#f4f4f4"
                border = "#999"
                label = ""

            boxes.append(
                f"""
                <div style="
                    width:70px;
                    height:70px;
                    display:flex;
                    flex-direction:column;
                    align-items:center;
                    justify-content:center;
                    border:3px solid {border};
                    background:{bg};
                    border-radius:10px;
                    margin:5px;
                    font-family:Arial;
                ">
                    <div style="font-size:26px;font-weight:bold">{value}</div>
                    <div style="font-size:11px">{label}</div>
                </div>
                """
            )

        if finished:

            status.value = "<b>Finished!</b> The list is sorted."

            explanation.value = (
                "<div style='margin-top:8px'><b>Bubble Sort:</b> "
                "Large values have bubbled to the right.</div>"
            )

        else:

            left = values[j]
            right = values[j + 1]

            status.value = (
                f"Pass <b>{pass_number}</b>. "
                f"Comparing index <b>{j}</b> and <b>{j + 1}</b>: "
                f"<b>{left}</b> and <b>{right}</b>."
            )

            if left > right:
                action = (
                    f"{left} &gt; {right} → "
                    "they will <b>swap</b>."
                )
            else:
                action = (
                    f"{left} ≤ {right} → "
                    "<b>no swap</b>."
                )

            explanation.value = (
                "<div style='margin-top:8px'>"
                + action +
                "</div>"
            )

        with output:

            clear_output(wait=True)

            display(
                HTML(
                    "<div style='display:flex;flex-wrap:wrap'>"
                    + "".join(boxes)
                    + "</div>"
                )
            )

    # -----------------------------
    # Reset state
    # -----------------------------
    def reset_state(new_values=None):

        nonlocal values
        nonlocal j
        nonlocal end
        nonlocal pass_number
        nonlocal finished
        nonlocal last_swapped
        nonlocal swapped_this_pass

        if new_values is not None:
            values = new_values[:]

        j = 0
        end = len(values) - 1
        pass_number = 1
        finished = False
        last_swapped = False
        swapped_this_pass = False

        render()

    # -----------------------------
    # Next comparison
    # -----------------------------
    def next_comparison(_=None):

        nonlocal j
        nonlocal end
        nonlocal pass_number
        nonlocal finished
        nonlocal last_swapped
        nonlocal swapped_this_pass

        if finished:
            return

        if values[j] > values[j + 1]:

            values[j], values[j + 1] = (
                values[j + 1],
                values[j]
            )

            last_swapped = True
            swapped_this_pass = True

        else:
            last_swapped = False

        j += 1

        if j >= end:

            if not swapped_this_pass:
                finished = True
                render()
                return

            end -= 1
            pass_number += 1
            j = 0
            swapped_this_pass = False

            if end <= 0:
                finished = True

        render()

    # -----------------------------
    # Finish pass
    # -----------------------------
    def finish_pass(_=None):

        nonlocal j
        nonlocal end
        nonlocal pass_number
        nonlocal finished
        nonlocal swapped_this_pass
        nonlocal last_swapped

        if finished:
            return

        while j < end:

            if values[j] > values[j + 1]:

                values[j], values[j + 1] = (
                    values[j + 1],
                    values[j]
                )

                swapped_this_pass = True

            j += 1

        if not swapped_this_pass:

            finished = True
            render()
            return

        end -= 1
        pass_number += 1
        j = 0
        swapped_this_pass = False
        last_swapped = False

        if end <= 0:
            finished = True

        render()

    # -----------------------------
    # Shuffle
    # -----------------------------
    def shuffle_values(_=None):

        nums = list(range(1, 9))
        random.shuffle(nums)

        reset_state(nums)

    # -----------------------------
    # Reset
    # -----------------------------
    def reset_values(_=None):

        reset_state(original_values)

    # -----------------------------
    # Connect buttons
    # -----------------------------
    next_btn.on_click(next_comparison)
    pass_btn.on_click(finish_pass)
    shuffle_btn.on_click(shuffle_values)
    reset_btn.on_click(reset_values)

    controls = widgets.HBox(
        [
            next_btn,
            pass_btn,
            shuffle_btn,
            reset_btn
        ]
    )

    display(
        title,
        status,
        controls,
        explanation,
        output
    )

    render()


create_bubble_sort_visualizer()
```

```python
import random
import math
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output


def create_bubble_planet_visualizer():

    # -----------------------------
    # Planet data
    # -----------------------------
    original_planets = [
        {"name": "Mercury", "diameter": 4879,   "symbol": "⚪"},
        {"name": "Venus",   "diameter": 12104,  "symbol": "🟡"},
        {"name": "Earth",   "diameter": 12756,  "symbol": "🌍"},
        {"name": "Mars",    "diameter": 6792,   "symbol": "🔴"},
        {"name": "Jupiter", "diameter": 142984, "symbol": "🟠"},
        {"name": "Saturn",  "diameter": 120536, "symbol": "🪐"},
        {"name": "Uranus",  "diameter": 51118,  "symbol": "🔷"},
        {"name": "Neptune", "diameter": 49528,  "symbol": "🔵"},
    ]

    planets = [p.copy() for p in original_planets]

    j = 0
    end = len(planets) - 1
    pass_number = 1
    finished = False
    swapped_this_pass = False

    # -----------------------------
    # UI
    # -----------------------------
    title = widgets.HTML(
        """
        <h3 style="margin-bottom:4px">Bubble Sort — Planet Size</h3>
        <div>
            Compare two neighboring planets by diameter.
            If the left planet is larger, swap them.
        </div>
        """
    )

    status = widgets.HTML()
    explanation = widgets.HTML()

    next_btn = widgets.Button(
        description="Next Comparison",
        button_style="primary"
    )

    pass_btn = widgets.Button(
        description="Finish This Pass"
    )

    shuffle_btn = widgets.Button(
        description="Shuffle"
    )

    reset_btn = widgets.Button(
        description="Reset"
    )

    output = widgets.Output()

    # -----------------------------
    # Helpers
    # -----------------------------
    def format_km(n):
        return f"{n:,} km"

    def visual_size(diameter):

        min_d = 4879
        max_d = 142984

        t = (
            math.sqrt(diameter) - math.sqrt(min_d)
        ) / (
            math.sqrt(max_d) - math.sqrt(min_d)
        )

        return 48 + t * 92

    def planet_html(planet, idx, role="", sorted_pos=False):

        size = visual_size(planet["diameter"])

        if sorted_pos:
            border = "#6aa66a"
            bg = "#e7f7e7"
            role_label = "sorted"

        elif role == "left":
            border = "#d6a600"
            bg = "#fff4c7"
            role_label = "left"

        elif role == "right":
            border = "#4a90e2"
            bg = "#e6f1ff"
            role_label = "right"

        else:
            border = "#aaaaaa"
            bg = "#f7f7f7"
            role_label = ""

        return f"""
        <div style="
            width:145px;
            min-height:285px;
            border:3px solid {border};
            background:{bg};
            border-radius:14px;
            margin:6px;
            padding:10px 6px 8px 6px;
            display:flex;
            flex-direction:column;
            align-items:center;
            justify-content:flex-end;
            box-sizing:border-box;
            font-family:Arial, sans-serif;
            text-align:center;
        ">
            <div style="
                font-size:17px;
                font-weight:bold;
                margin-bottom:3px;
            ">
                {planet["name"]}
            </div>

            <div style="
                font-size:13px;
                margin-bottom:8px;
            ">
                Diameter: {format_km(planet["diameter"])}
            </div>

            <div style="
                width:{size}px;
                height:{size}px;
                border-radius:50%;
                display:flex;
                align-items:center;
                justify-content:center;
                font-size:{max(28, size * 0.48)}px;
                line-height:1;
                margin:auto 0 10px 0;
            ">
                {planet["symbol"]}
            </div>

            <div style="
                font-size:11px;
                min-height:18px;
                font-weight:bold;
            ">
                {role_label}
            </div>

            <div style="
                margin-top:2px;
                font-size:11px;
                color:#666;
            ">
                index {idx}
            </div>
        </div>
        """

    # -----------------------------
    # Render
    # -----------------------------
    def render():

        cards = []

        for idx, planet in enumerate(planets):

            if finished or idx > end:
                cards.append(
                    planet_html(
                        planet,
                        idx,
                        sorted_pos=True
                    )
                )

            elif idx == j:
                cards.append(
                    planet_html(
                        planet,
                        idx,
                        role="left"
                    )
                )

            elif idx == j + 1:
                cards.append(
                    planet_html(
                        planet,
                        idx,
                        role="right"
                    )
                )

            else:
                cards.append(
                    planet_html(
                        planet,
                        idx
                    )
                )

        if finished:

            status.value = (
                "<b>Finished!</b> "
                "The planets are arranged from smallest diameter "
                "to largest diameter."
            )

            explanation.value = (
                "<div style='margin-top:8px'>"
                "<b>What happened?</b> "
                "During each pass, the largest unsorted planet "
                "moved toward the right one comparison at a time."
                "</div>"
            )

        else:

            left = planets[j]
            right = planets[j + 1]

            status.value = (
                f"Pass <b>{pass_number}</b>. "
                f"Comparing index <b>{j}</b> and <b>{j + 1}</b>: "
                f"<b>{left['name']}</b> "
                f"({format_km(left['diameter'])}) and "
                f"<b>{right['name']}</b> "
                f"({format_km(right['diameter'])})."
            )

            if left["diameter"] > right["diameter"]:

                action = (
                    f"{left['name']} is larger than "
                    f"{right['name']} → "
                    "they will <b>swap</b>."
                )

            else:

                action = (
                    f"{left['name']} is not larger than "
                    f"{right['name']} → "
                    "<b>no swap</b>."
                )

            explanation.value = (
                f"<div style='margin-top:8px'>"
                f"{action}"
                f"</div>"
            )

        legend = """
        <div style="
            display:flex;
            gap:18px;
            flex-wrap:wrap;
            margin:8px 0 2px 0;
            font-family:Arial, sans-serif;
            font-size:13px;
        ">
            <span>🟨 left planet</span>
            <span>🟦 right planet</span>
            <span>🟩 final position</span>
        </div>
        """

        goal = """
        <div style="
            margin-top:8px;
            padding:8px 10px;
            border-left:4px solid #888;
            background:#fafafa;
            font-family:Arial, sans-serif;
        ">
            <b>Starting order:</b>
            Distance from the Sun — nearest to farthest.<br>
            Mercury → Venus → Earth → Mars →
            Jupiter → Saturn → Uranus → Neptune

            <br><br>

            <b>Sorting goal:</b>
            Rearrange the planets from smallest diameter
            to largest diameter.

            <br>

            <span style="
                font-size:12px;
                color:#666;
            ">
                Note: visual planet sizes are compressed
                so every planet stays visible.
                The algorithm compares the diameter numbers.
            </span>
        </div>
        """

        with output:

            clear_output(wait=True)

            display(
                HTML(
                    goal
                    + legend
                    + "<div style='display:flex;"
                    "align-items:flex-end;"
                    "flex-wrap:wrap;'>"
                    + "".join(cards)
                    + "</div>"
                )
            )

    # -----------------------------
    # Reset state
    # -----------------------------
    def reset_state(new_planets=None):

        nonlocal planets
        nonlocal j
        nonlocal end
        nonlocal pass_number
        nonlocal finished
        nonlocal swapped_this_pass

        if new_planets is not None:
            planets = [
                p.copy()
                for p in new_planets
            ]

        j = 0
        end = len(planets) - 1
        pass_number = 1
        finished = False
        swapped_this_pass = False

        render()

    # -----------------------------
    # Next comparison
    # -----------------------------
    def next_comparison(_=None):

        nonlocal j
        nonlocal end
        nonlocal pass_number
        nonlocal finished
        nonlocal swapped_this_pass

        if finished:
            return

        if (
            planets[j]["diameter"]
            >
            planets[j + 1]["diameter"]
        ):

            planets[j], planets[j + 1] = (
                planets[j + 1],
                planets[j]
            )

            swapped_this_pass = True

        j += 1

        if j >= end:

            if not swapped_this_pass:

                finished = True
                render()
                return

            end -= 1
            pass_number += 1
            j = 0
            swapped_this_pass = False

            if end <= 0:
                finished = True

        render()

    # -----------------------------
    # Finish pass
    # -----------------------------
    def finish_pass(_=None):

        nonlocal j
        nonlocal end
        nonlocal pass_number
        nonlocal finished
        nonlocal swapped_this_pass

        if finished:
            return

        while j < end:

            if (
                planets[j]["diameter"]
                >
                planets[j + 1]["diameter"]
            ):

                planets[j], planets[j + 1] = (
                    planets[j + 1],
                    planets[j]
                )

                swapped_this_pass = True

            j += 1

        if not swapped_this_pass:

            finished = True
            render()
            return

        end -= 1
        pass_number += 1
        j = 0
        swapped_this_pass = False

        if end <= 0:
            finished = True

        render()

    # -----------------------------
    # Shuffle
    # -----------------------------
    def shuffle_values(_=None):

        new_planets = [
            p.copy()
            for p in original_planets
        ]

        random.shuffle(new_planets)

        reset_state(new_planets)

    # -----------------------------
    # Reset
    # -----------------------------
    def reset_values(_=None):

        reset_state(original_planets)

    # -----------------------------
    # Connect buttons
    # -----------------------------
    next_btn.on_click(next_comparison)
    pass_btn.on_click(finish_pass)
    shuffle_btn.on_click(shuffle_values)
    reset_btn.on_click(reset_values)

    controls = widgets.HBox(
        [
            next_btn,
            pass_btn,
            shuffle_btn,
            reset_btn
        ]
    )

    display(
        title,
        status,
        controls,
        explanation,
        output
    )

    render()


create_bubble_planet_visualizer()
```

```python
import random
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output


def create_selection_sort_visualizer():

    # -----------------------------
    # State
    # -----------------------------
    original_values = [6, 3, 8, 2, 7, 1, 5, 4]
    values = original_values[:]

    i = 0
    j = 1
    min_index = 0
    finished = False

    # -----------------------------
    # UI
    # -----------------------------
    title = widgets.HTML(
        "<h3 style='margin-bottom:4px'>Selection Sort Visualizer</h3>"
        "<div>Find the smallest value in the unsorted part, "
        "then swap it into place.</div>"
    )

    status = widgets.HTML()

    next_btn = widgets.Button(
        description="Next Comparison",
        button_style="primary"
    )

    pass_btn = widgets.Button(
        description="Finish This Pass"
    )

    shuffle_btn = widgets.Button(
        description="Shuffle"
    )

    reset_btn = widgets.Button(
        description="Reset"
    )

    output = widgets.Output()

    # -----------------------------
    # Render
    # -----------------------------
    def render():

        boxes = []

        for idx, value in enumerate(values):

            if finished or idx < i:
                bg = "#dff5df"
                border = "#6aa66a"
                label = "sorted"

            elif idx == min_index:
                bg = "#fff0b3"
                border = "#d6a600"
                label = "minimum"

            elif idx == j:
                bg = "#dcecff"
                border = "#4a90e2"
                label = "compare"

            else:
                bg = "#f4f4f4"
                border = "#999"
                label = ""

            boxes.append(
                f"""
                <div style="
                    width:70px;
                    height:70px;
                    display:flex;
                    flex-direction:column;
                    align-items:center;
                    justify-content:center;
                    border:3px solid {border};
                    background:{bg};
                    border-radius:10px;
                    margin:5px;
                    font-family:Arial;
                ">
                    <div style="
                        font-size:26px;
                        font-weight:bold;
                    ">
                        {value}
                    </div>

                    <div style="
                        font-size:11px;
                    ">
                        {label}
                    </div>
                </div>
                """
            )

        if finished:

            status.value = (
                "<b>Finished!</b> The list is sorted."
            )

        else:

            status.value = (
                f"Pass starts at index <b>{i}</b>. "
                f"Current minimum = <b>{values[min_index]}</b> "
                f"(index {min_index}). "
                f"Comparing with index <b>{j}</b>."
            )

        with output:

            clear_output(wait=True)

            display(
                HTML(
                    "<div style='display:flex;flex-wrap:wrap'>"
                    + "".join(boxes)
                    + "</div>"
                )
            )

    # -----------------------------
    # Reset state
    # -----------------------------
    def reset_state(new_values=None):

        nonlocal values
        nonlocal i
        nonlocal j
        nonlocal min_index
        nonlocal finished

        if new_values is not None:
            values = new_values[:]

        i = 0
        min_index = 0
        j = 1
        finished = False

        render()

    # -----------------------------
    # Next comparison
    # -----------------------------
    def advance_one_comparison(_=None):

        nonlocal i
        nonlocal j
        nonlocal min_index
        nonlocal finished

        if finished:
            return

        n = len(values)

        if j < n:

            if values[j] < values[min_index]:
                min_index = j

            j += 1

        # End of this pass
        if j >= n:

            values[i], values[min_index] = (
                values[min_index],
                values[i]
            )

            i += 1

            if i >= n - 1:

                finished = True
                render()
                return

            min_index = i
            j = i + 1

        render()

    # -----------------------------
    # Finish this pass
    # -----------------------------
    def finish_pass(_=None):

        nonlocal i
        nonlocal j
        nonlocal min_index
        nonlocal finished

        if finished:
            return

        n = len(values)

        while j < n:

            if values[j] < values[min_index]:
                min_index = j

            j += 1

        values[i], values[min_index] = (
            values[min_index],
            values[i]
        )

        i += 1

        if i >= n - 1:

            finished = True

        else:

            min_index = i
            j = i + 1

        render()

    # -----------------------------
    # Shuffle
    # -----------------------------
    def shuffle_values(_=None):

        nums = list(range(1, 9))
        random.shuffle(nums)

        reset_state(nums)

    # -----------------------------
    # Reset
    # -----------------------------
    def reset_values(_=None):

        reset_state(original_values)

    # -----------------------------
    # Connect buttons
    # -----------------------------
    next_btn.on_click(advance_one_comparison)
    pass_btn.on_click(finish_pass)
    shuffle_btn.on_click(shuffle_values)
    reset_btn.on_click(reset_values)

    controls = widgets.HBox(
        [
            next_btn,
            pass_btn,
            shuffle_btn,
            reset_btn
        ]
    )

    display(
        title,
        status,
        controls,
        output
    )

    render()


create_selection_sort_visualizer()
```

```python
import random
import math
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output


def create_selection_planet_visualizer():

    # -----------------------------
    # Planet data
    # -----------------------------
    original_planets = [
        {"name": "Mercury", "diameter": 4879,   "symbol": "⚪"},
        {"name": "Venus",   "diameter": 12104,  "symbol": "🟡"},
        {"name": "Earth",   "diameter": 12756,  "symbol": "🌍"},
        {"name": "Mars",    "diameter": 6792,   "symbol": "🔴"},
        {"name": "Jupiter", "diameter": 142984, "symbol": "🟠"},
        {"name": "Saturn",  "diameter": 120536, "symbol": "🪐"},
        {"name": "Uranus",  "diameter": 51118,  "symbol": "🔷"},
        {"name": "Neptune", "diameter": 49528,  "symbol": "🔵"},
    ]

    planets = [p.copy() for p in original_planets]

    i = 0
    j = 1
    min_index = 0
    finished = False

    # -----------------------------
    # UI
    # -----------------------------
    title = widgets.HTML(
        """
        <h3 style="margin-bottom:4px">Selection Sort — Planet Size</h3>
        <div>
            Find the smallest planet in the unsorted group.
            After checking all remaining planets, swap it into the first unsorted position.
        </div>
        """
    )

    status = widgets.HTML()
    explanation = widgets.HTML()

    next_btn = widgets.Button(
        description="Next Comparison",
        button_style="primary"
    )

    pass_btn = widgets.Button(
        description="Finish This Pass"
    )

    shuffle_btn = widgets.Button(
        description="Shuffle"
    )

    reset_btn = widgets.Button(
        description="Reset"
    )

    output = widgets.Output()

    # -----------------------------
    # Helpers
    # -----------------------------
    def format_km(n):
        return f"{n:,} km"

    def visual_size(diameter):

        min_d = 4879
        max_d = 142984

        t = (
            math.sqrt(diameter) - math.sqrt(min_d)
        ) / (
            math.sqrt(max_d) - math.sqrt(min_d)
        )

        return 48 + t * 92

    def planet_html(planet, idx, role="", sorted_pos=False):

        size = visual_size(planet["diameter"])

        if sorted_pos:
            border = "#6aa66a"
            bg = "#e7f7e7"
            role_label = "sorted"

        elif role == "minimum":
            border = "#d6a600"
            bg = "#fff4c7"
            role_label = "smallest so far"

        elif role == "compare":
            border = "#4a90e2"
            bg = "#e6f1ff"
            role_label = "compare"

        else:
            border = "#aaaaaa"
            bg = "#f7f7f7"
            role_label = ""

        return f"""
        <div style="
            width:145px;
            min-height:290px;
            border:3px solid {border};
            background:{bg};
            border-radius:14px;
            margin:6px;
            padding:10px 6px 8px 6px;
            display:flex;
            flex-direction:column;
            align-items:center;
            justify-content:flex-end;
            box-sizing:border-box;
            font-family:Arial, sans-serif;
            text-align:center;
        ">

            <div style="
                font-size:17px;
                font-weight:bold;
                margin-bottom:3px;
            ">
                {planet["name"]}
            </div>

            <div style="
                font-size:13px;
                margin-bottom:8px;
            ">
                Diameter: {format_km(planet["diameter"])}
            </div>

            <div style="
                width:{size}px;
                height:{size}px;
                border-radius:50%;
                display:flex;
                align-items:center;
                justify-content:center;
                font-size:{max(28, size * 0.48)}px;
                line-height:1;
                margin:auto 0 10px 0;
            ">
                {planet["symbol"]}
            </div>

            <div style="
                font-size:11px;
                min-height:30px;
                font-weight:bold;
            ">
                {role_label}
            </div>

            <div style="
                margin-top:2px;
                font-size:11px;
                color:#666;
            ">
                index {idx}
            </div>

        </div>
        """

    # -----------------------------
    # Render
    # -----------------------------
    def render():

        cards = []

        for idx, planet in enumerate(planets):

            if finished or idx < i:
                cards.append(
                    planet_html(
                        planet,
                        idx,
                        sorted_pos=True
                    )
                )

            elif idx == min_index:
                cards.append(
                    planet_html(
                        planet,
                        idx,
                        role="minimum"
                    )
                )

            elif idx == j:
                cards.append(
                    planet_html(
                        planet,
                        idx,
                        role="compare"
                    )
                )

            else:
                cards.append(
                    planet_html(
                        planet,
                        idx
                    )
                )

        if finished:

            status.value = (
                "<b>Finished!</b> "
                "The planets are arranged from smallest diameter "
                "to largest diameter."
            )

            explanation.value = (
                "<div style='margin-top:8px'>"
                "<b>Selection Sort:</b> "
                "Each pass searched all remaining planets, "
                "remembered the smallest one, and then made one swap."
                "</div>"
            )

        else:

            current_min = planets[min_index]
            current = planets[j]

            status.value = (
                f"Pass starts at index <b>{i}</b>. "
                f"Smallest so far: <b>{current_min['name']}</b> "
                f"({format_km(current_min['diameter'])}). "
                f"Comparing with <b>{current['name']}</b> "
                f"({format_km(current['diameter'])})."
            )

            if current["diameter"] < current_min["diameter"]:

                action = (
                    f"{current['name']} is smaller than "
                    f"{current_min['name']} → "
                    f"<b>{current['name']} becomes the new minimum.</b>"
                )

            else:

                action = (
                    f"{current['name']} is not smaller than "
                    f"{current_min['name']} → "
                    f"keep <b>{current_min['name']}</b> as the minimum."
                )

            explanation.value = (
                f"<div style='margin-top:8px'>"
                f"{action}"
                f"</div>"
            )

        legend = """
        <div style="
            display:flex;
            gap:18px;
            flex-wrap:wrap;
            margin:8px 0 2px 0;
            font-family:Arial, sans-serif;
            font-size:13px;
        ">
            <span>🟨 smallest so far</span>
            <span>🟦 planet being compared</span>
            <span>🟩 final position</span>
        </div>
        """

        goal = """
        <div style="
            margin-top:8px;
            padding:8px 10px;
            border-left:4px solid #888;
            background:#fafafa;
            font-family:Arial, sans-serif;
        ">
            <b>Starting order:</b>
            Distance from the Sun — nearest to farthest.<br>

            Mercury → Venus → Earth → Mars →
            Jupiter → Saturn → Uranus → Neptune

            <br><br>

            <b>Sorting goal:</b>
            Rearrange the planets from smallest diameter
            to largest diameter.

            <br>

            <span style="
                font-size:12px;
                color:#666;
            ">
                Visual planet sizes are compressed so every planet remains visible.
                Selection Sort compares the diameter numbers.
            </span>
        </div>
        """

        with output:

            clear_output(wait=True)

            display(
                HTML(
                    goal
                    + legend
                    + "<div style='display:flex;"
                    "align-items:flex-end;"
                    "flex-wrap:wrap;'>"
                    + "".join(cards)
                    + "</div>"
                )
            )

    # -----------------------------
    # Reset state
    # -----------------------------
    def reset_state(new_planets=None):

        nonlocal planets
        nonlocal i
        nonlocal j
        nonlocal min_index
        nonlocal finished

        if new_planets is not None:
            planets = [
                p.copy()
                for p in new_planets
            ]

        i = 0
        min_index = 0
        j = 1
        finished = False

        render()

    # -----------------------------
    # Next comparison
    # -----------------------------
    def next_comparison(_=None):

        nonlocal i
        nonlocal j
        nonlocal min_index
        nonlocal finished

        if finished:
            return

        n = len(planets)

        if (
            planets[j]["diameter"]
            <
            planets[min_index]["diameter"]
        ):
            min_index = j

        j += 1

        if j >= n:

            planets[i], planets[min_index] = (
                planets[min_index],
                planets[i]
            )

            i += 1

            if i >= n - 1:

                finished = True
                render()
                return

            min_index = i
            j = i + 1

        render()

    # -----------------------------
    # Finish pass
    # -----------------------------
    def finish_pass(_=None):

        nonlocal i
        nonlocal j
        nonlocal min_index
        nonlocal finished

        if finished:
            return

        n = len(planets)

        while j < n:

            if (
                planets[j]["diameter"]
                <
                planets[min_index]["diameter"]
            ):
                min_index = j

            j += 1

        planets[i], planets[min_index] = (
            planets[min_index],
            planets[i]
        )

        i += 1

        if i >= n - 1:

            finished = True

        else:

            min_index = i
            j = i + 1

        render()

    # -----------------------------
    # Shuffle
    # -----------------------------
    def shuffle_values(_=None):

        new_planets = [
            p.copy()
            for p in original_planets
        ]

        random.shuffle(new_planets)

        reset_state(new_planets)

    # -----------------------------
    # Reset
    # -----------------------------
    def reset_values(_=None):

        reset_state(original_planets)

    # -----------------------------
    # Connect buttons
    # -----------------------------
    next_btn.on_click(next_comparison)
    pass_btn.on_click(finish_pass)
    shuffle_btn.on_click(shuffle_values)
    reset_btn.on_click(reset_values)

    controls = widgets.HBox(
        [
            next_btn,
            pass_btn,
            shuffle_btn,
            reset_btn
        ]
    )

    display(
        title,
        status,
        controls,
        explanation,
        output
    )

    render()


create_selection_planet_visualizer()
```

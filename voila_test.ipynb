### Bubble Sort Flow ###

```python
# Bubble Sort Visualizer for Jupyter Notebook
# Run this entire code in one Jupyter cell.
# Requires: ipywidgets

import random
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# -----------------------------
# State
# -----------------------------
original_values = [6, 3, 8, 2, 7, 1, 5, 4]
values = original_values[:]
j = 0                       # left element of current adjacent pair
end = len(values) - 1       # last unsorted index
pass_number = 1
finished = False
last_swapped = False
swapped_this_pass = False

# -----------------------------
# UI
# -----------------------------
title = widgets.HTML(
    "<h3 style='margin-bottom:4px'>Bubble Sort Visualizer</h3>"
    "<div>Compare adjacent values. If the left value is bigger, swap them.</div>"
)

status = widgets.HTML()
explanation = widgets.HTML()

next_btn = widgets.Button(description="Next Comparison", button_style="primary")
pass_btn = widgets.Button(description="Finish This Pass")
shuffle_btn = widgets.Button(description="Shuffle")
reset_btn = widgets.Button(description="Reset")

output = widgets.Output()

def render():
    global values, j, end, pass_number, finished, last_swapped

    boxes = []

    for idx, value in enumerate(values):
        if finished or idx > end:
            bg = "#dff5df"
            border = "#6aa66a"
            label = "sorted"

        elif idx == j:
            bg = "#fff0b3"
            border = "#d6a600"
            label = "left"

        elif idx == j + 1:
            bg = "#dcecff"
            border = "#4a90e2"
            label = "right"

        else:
            bg = "#f4f4f4"
            border = "#999"
            label = ""

        boxes.append(
            f"""
            <div style="
                width:70px;
                height:70px;
                display:flex;
                flex-direction:column;
                align-items:center;
                justify-content:center;
                border:3px solid {border};
                background:{bg};
                border-radius:10px;
                margin:5px;
                font-family:Arial;
            ">
                <div style="font-size:26px;font-weight:bold">{value}</div>
                <div style="font-size:11px">{label}</div>
            </div>
            """
        )

    if finished:
        status.value = "<b>Finished!</b> The list is sorted."
        explanation.value = (
            "<div style='margin-top:8px'><b>Bubble Sort:</b> "
            "Large values have bubbled to the right.</div>"
        )
    else:
        left = values[j]
        right = values[j + 1]

        status.value = (
            f"Pass <b>{pass_number}</b>. "
            f"Comparing index <b>{j}</b> and <b>{j + 1}</b>: "
            f"<b>{left}</b> and <b>{right}</b>."
        )

        if left > right:
            action = f"{left} &gt; {right} → they will <b>swap</b>."
        else:
            action = f"{left} ≤ {right} → <b>no swap</b>."

        explanation.value = (
            "<div style='margin-top:8px'>"
            + action +
            "</div>"
        )

    with output:
        clear_output(wait=True)
        display(HTML(
            "<div style='display:flex;flex-wrap:wrap'>"
            + "".join(boxes)
            + "</div>"
        ))

def reset_state(new_values=None):
    global values, j, end, pass_number, finished
    global last_swapped, swapped_this_pass

    if new_values is not None:
        values = new_values[:]

    j = 0
    end = len(values) - 1
    pass_number = 1
    finished = False
    last_swapped = False
    swapped_this_pass = False
    render()

def next_comparison(_=None):
    global values, j, end, pass_number, finished
    global last_swapped, swapped_this_pass

    if finished:
        return

    # Compare adjacent pair and swap immediately if needed
    if values[j] > values[j + 1]:
        values[j], values[j + 1] = values[j + 1], values[j]
        last_swapped = True
        swapped_this_pass = True
    else:
        last_swapped = False

    j += 1

    # End of one pass
    if j >= end:
        # The value at 'end' is now in final position
        if not swapped_this_pass:
            finished = True
            render()
            return

        end -= 1
        pass_number += 1
        j = 0
        swapped_this_pass = False

        if end <= 0:
            finished = True

    render()

def finish_pass(_=None):
    global values, j, end, pass_number, finished
    global swapped_this_pass, last_swapped

    if finished:
        return

    while j < end:
        if values[j] > values[j + 1]:
            values[j], values[j + 1] = values[j + 1], values[j]
            swapped_this_pass = True
        j += 1

    if not swapped_this_pass:
        finished = True
        render()
        return

    end -= 1
    pass_number += 1
    j = 0
    swapped_this_pass = False
    last_swapped = False

    if end <= 0:
        finished = True

    render()

def shuffle_values(_=None):
    nums = list(range(1, 9))
    random.shuffle(nums)
    reset_state(nums)

def reset_values(_=None):
    reset_state(original_values)

next_btn.on_click(next_comparison)
pass_btn.on_click(finish_pass)
shuffle_btn.on_click(shuffle_values)
reset_btn.on_click(reset_values)

controls = widgets.HBox([next_btn, pass_btn, shuffle_btn, reset_btn])

display(title, status, controls, explanation, output)
render()
```

```python
# Bubble Sort Height-Order Visualizer — US Feet & Inches for Jupyter Notebook
# Run this entire code in one Jupyter cell.
# Requires: ipywidgets

import random
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# -----------------------------
# State
# -----------------------------
original_heights = [62, 56, 67, 51, 64, 69, 57, 63]  # total inches
heights = original_heights[:]

j = 0
end = len(heights) - 1
pass_number = 1
finished = False
swapped_this_pass = False

# -----------------------------
# UI
# -----------------------------
title = widgets.HTML(
    """
    <h3 style="margin-bottom:4px">Bubble Sort — Height Order</h3>
    <div>Compare two students standing next to each other.
    If the left student is taller, they swap places.</div>
    """
)

status = widgets.HTML()
explanation = widgets.HTML()

next_btn = widgets.Button(description="Next Comparison", button_style="primary")
pass_btn = widgets.Button(description="Finish This Pass")
shuffle_btn = widgets.Button(description="Shuffle")
reset_btn = widgets.Button(description="Reset")

output = widgets.Output()

def format_height(total_inches):
    feet = total_inches // 12
    inches = total_inches % 12
    return f"{feet}' {inches}\""

def display_height(total_inches):
    cm = round(total_inches * 2.54)
    return f"{format_height(total_inches)} ({cm} cm)"

def person_html(height_cm, idx, role="", sorted_pos=False):
    """
    Draw a simple person using only HTML/CSS.
    Visual height is scaled from the student's height in cm.
    """
    # Scale visual body height into a comfortable notebook range
    visual_h = 110 + (height_cm - 50) * 5.0

    if sorted_pos:
        border = "#6aa66a"
        bg = "#e7f7e7"
        role_label = "sorted"
    elif role == "left":
        border = "#d6a600"
        bg = "#fff4c7"
        role_label = "left"
    elif role == "right":
        border = "#4a90e2"
        bg = "#e6f1ff"
        role_label = "right"
    else:
        border = "#aaaaaa"
        bg = "#f7f7f7"
        role_label = ""

    body_h = max(45, visual_h - 62)
    leg_h = max(22, body_h * 0.34)
    torso_h = body_h - leg_h

    return f"""
    <div style="
        width:105px;
        min-height:265px;
        border:3px solid {border};
        background:{bg};
        border-radius:12px;
        margin:6px;
        padding:8px 6px 6px 6px;
        display:flex;
        flex-direction:column;
        align-items:center;
        justify-content:flex-end;
        box-sizing:border-box;
        font-family:Arial, sans-serif;
    ">
        <div style="
            font-size:17px;
            font-weight:bold;
            margin-bottom:4px;
        ">{display_height(height_cm)}</div>

        <div style="
            height:{visual_h}px;
            width:58px;
            display:flex;
            flex-direction:column;
            align-items:center;
            justify-content:flex-end;
        ">
            <!-- head -->
            <div style="
                width:34px;
                height:34px;
                border-radius:50%;
                background:#f2c7a5;
                border:2px solid #555;
                box-sizing:border-box;
                flex-shrink:0;
            "></div>

            <!-- torso -->
            <div style="
                width:38px;
                height:{torso_h}px;
                min-height:38px;
                background:#7aa7d9;
                border:2px solid #555;
                border-radius:8px 8px 5px 5px;
                box-sizing:border-box;
                flex-shrink:0;
            "></div>

            <!-- legs -->
            <div style="
                width:38px;
                height:{leg_h}px;
                display:flex;
                gap:8px;
                justify-content:center;
                flex-shrink:0;
            ">
                <div style="
                    width:8px;
                    height:100%;
                    background:#555;
                    border-radius:3px;
                "></div>
                <div style="
                    width:8px;
                    height:100%;
                    background:#555;
                    border-radius:3px;
                "></div>
            </div>
        </div>

        <div style="
            margin-top:5px;
            font-size:12px;
            min-height:16px;
            font-weight:bold;
        ">{role_label}</div>

        <div style="
            margin-top:2px;
            font-size:11px;
            color:#666;
        ">index {idx}</div>
    </div>
    """

def render():
    global heights, j, end, pass_number, finished

    cards = []

    for idx, h in enumerate(heights):
        if finished or idx > end:
            cards.append(person_html(h, idx, sorted_pos=True))
        elif idx == j:
            cards.append(person_html(h, idx, role="left"))
        elif idx == j + 1:
            cards.append(person_html(h, idx, role="right"))
        else:
            cards.append(person_html(h, idx))

    if finished:
        status.value = "<b>Finished!</b> Everyone is arranged from shortest to tallest."
        explanation.value = (
            "<div style='margin-top:8px'>"
            "<b>What happened?</b> The tallest unsorted student moved toward the right "
            "one comparison at a time during each pass."
            "</div>"
        )
    else:
        left = heights[j]
        right = heights[j + 1]

        status.value = (
            f"Pass <b>{pass_number}</b>. "
            f"Comparing index <b>{j}</b> and <b>{j + 1}</b>: "
            f"<b>{display_height(left)}</b> and <b>{display_height(right)}</b>."
        )

        if left > right:
            action = (
                f"The left student is taller: "
                f"<b>{display_height(left)} &gt; {display_height(right)}</b> → they will <b>swap</b>."
            )
        else:
            action = (
                f"The left student is not taller: "
                f"<b>{display_height(left)} ≤ {display_height(right)}</b> → <b>no swap</b>."
            )

        explanation.value = f"<div style='margin-top:8px'>{action}</div>"

    legend = """
    <div style="
        display:flex;
        gap:18px;
        flex-wrap:wrap;
        margin:8px 0 2px 0;
        font-family:Arial, sans-serif;
        font-size:13px;
    ">
        <span>🟨 left student</span>
        <span>🟦 right student</span>
        <span>🟩 final position</span>
    </div>
    """

    goal = """
    <div style="
        margin-top:8px;
        padding:8px 10px;
        border-left:4px solid #888;
        background:#fafafa;
        font-family:Arial, sans-serif;
    ">
        <b>Goal:</b> Arrange the students from shortest to tallest.
    </div>
    """

    with output:
        clear_output(wait=True)
        display(HTML(
            goal
            + legend
            + "<div style='display:flex;align-items:flex-end;flex-wrap:wrap;'>"
            + "".join(cards)
            + "</div>"
        ))

def reset_state(new_heights=None):
    global heights, j, end, pass_number, finished, swapped_this_pass

    if new_heights is not None:
        heights = new_heights[:]

    j = 0
    end = len(heights) - 1
    pass_number = 1
    finished = False
    swapped_this_pass = False
    render()

def next_comparison(_=None):
    global heights, j, end, pass_number, finished, swapped_this_pass

    if finished:
        return

    if heights[j] > heights[j + 1]:
        heights[j], heights[j + 1] = heights[j + 1], heights[j]
        swapped_this_pass = True

    j += 1

    if j >= end:
        if not swapped_this_pass:
            finished = True
            render()
            return

        end -= 1
        pass_number += 1
        j = 0
        swapped_this_pass = False

        if end <= 0:
            finished = True

    render()

def finish_pass(_=None):
    global heights, j, end, pass_number, finished, swapped_this_pass

    if finished:
        return

    while j < end:
        if heights[j] > heights[j + 1]:
            heights[j], heights[j + 1] = heights[j + 1], heights[j]
            swapped_this_pass = True
        j += 1

    if not swapped_this_pass:
        finished = True
        render()
        return

    end -= 1
    pass_number += 1
    j = 0
    swapped_this_pass = False

    if end <= 0:
        finished = True

    render()

def shuffle_values(_=None):
    new_heights = original_heights[:]
    random.shuffle(new_heights)
    reset_state(new_heights)

def reset_values(_=None):
    reset_state(original_heights)

next_btn.on_click(next_comparison)
pass_btn.on_click(finish_pass)
shuffle_btn.on_click(shuffle_values)
reset_btn.on_click(reset_values)

controls = widgets.HBox([next_btn, pass_btn, shuffle_btn, reset_btn])

display(title, status, controls, explanation, output)
render()
```

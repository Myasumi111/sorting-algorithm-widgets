## Bubble Sort Flow ##

```python
# Bubble Sort Visualizer for Jupyter Notebook
# Run this entire code in one Jupyter cell.
# Requires: ipywidgets

import random
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# -----------------------------
# State
# -----------------------------
original_values = [6, 3, 8, 2, 7, 1, 5, 4]
values = original_values[:]
j = 0                       # left element of current adjacent pair
end = len(values) - 1       # last unsorted index
pass_number = 1
finished = False
last_swapped = False
swapped_this_pass = False

# -----------------------------
# UI
# -----------------------------
title = widgets.HTML(
    "<h3 style='margin-bottom:4px'>Bubble Sort Visualizer</h3>"
    "<div>Compare adjacent values. If the left value is bigger, swap them.</div>"
)

status = widgets.HTML()
explanation = widgets.HTML()

next_btn = widgets.Button(description="Next Comparison", button_style="primary")
pass_btn = widgets.Button(description="Finish This Pass")
shuffle_btn = widgets.Button(description="Shuffle")
reset_btn = widgets.Button(description="Reset")

output = widgets.Output()

def render():
    global values, j, end, pass_number, finished, last_swapped

    boxes = []

    for idx, value in enumerate(values):
        if finished or idx > end:
            bg = "#dff5df"
            border = "#6aa66a"
            label = "sorted"

        elif idx == j:
            bg = "#fff0b3"
            border = "#d6a600"
            label = "left"

        elif idx == j + 1:
            bg = "#dcecff"
            border = "#4a90e2"
            label = "right"

        else:
            bg = "#f4f4f4"
            border = "#999"
            label = ""

        boxes.append(
            f"""
            <div style="
                width:70px;
                height:70px;
                display:flex;
                flex-direction:column;
                align-items:center;
                justify-content:center;
                border:3px solid {border};
                background:{bg};
                border-radius:10px;
                margin:5px;
                font-family:Arial;
            ">
                <div style="font-size:26px;font-weight:bold">{value}</div>
                <div style="font-size:11px">{label}</div>
            </div>
            """
        )

    if finished:
        status.value = "<b>Finished!</b> The list is sorted."
        explanation.value = (
            "<div style='margin-top:8px'><b>Bubble Sort:</b> "
            "Large values have bubbled to the right.</div>"
        )
    else:
        left = values[j]
        right = values[j + 1]

        status.value = (
            f"Pass <b>{pass_number}</b>. "
            f"Comparing index <b>{j}</b> and <b>{j + 1}</b>: "
            f"<b>{left}</b> and <b>{right}</b>."
        )

        if left > right:
            action = f"{left} &gt; {right} → they will <b>swap</b>."
        else:
            action = f"{left} ≤ {right} → <b>no swap</b>."

        explanation.value = (
            "<div style='margin-top:8px'>"
            + action +
            "</div>"
        )

    with output:
        clear_output(wait=True)
        display(HTML(
            "<div style='display:flex;flex-wrap:wrap'>"
            + "".join(boxes)
            + "</div>"
        ))

def reset_state(new_values=None):
    global values, j, end, pass_number, finished
    global last_swapped, swapped_this_pass

    if new_values is not None:
        values = new_values[:]

    j = 0
    end = len(values) - 1
    pass_number = 1
    finished = False
    last_swapped = False
    swapped_this_pass = False
    render()

def next_comparison(_=None):
    global values, j, end, pass_number, finished
    global last_swapped, swapped_this_pass

    if finished:
        return

    # Compare adjacent pair and swap immediately if needed
    if values[j] > values[j + 1]:
        values[j], values[j + 1] = values[j + 1], values[j]
        last_swapped = True
        swapped_this_pass = True
    else:
        last_swapped = False

    j += 1

    # End of one pass
    if j >= end:
        # The value at 'end' is now in final position
        if not swapped_this_pass:
            finished = True
            render()
            return

        end -= 1
        pass_number += 1
        j = 0
        swapped_this_pass = False

        if end <= 0:
            finished = True

    render()

def finish_pass(_=None):
    global values, j, end, pass_number, finished
    global swapped_this_pass, last_swapped

    if finished:
        return

    while j < end:
        if values[j] > values[j + 1]:
            values[j], values[j + 1] = values[j + 1], values[j]
            swapped_this_pass = True
        j += 1

    if not swapped_this_pass:
        finished = True
        render()
        return

    end -= 1
    pass_number += 1
    j = 0
    swapped_this_pass = False
    last_swapped = False

    if end <= 0:
        finished = True

    render()

def shuffle_values(_=None):
    nums = list(range(1, 9))
    random.shuffle(nums)
    reset_state(nums)

def reset_values(_=None):
    reset_state(original_values)

next_btn.on_click(next_comparison)
pass_btn.on_click(finish_pass)
shuffle_btn.on_click(shuffle_values)
reset_btn.on_click(reset_values)

controls = widgets.HBox([next_btn, pass_btn, shuffle_btn, reset_btn])

display(title, status, controls, explanation, output)
render()
```

```python
# Bubble Sort Height-Order Visualizer — US Feet & Inches for Jupyter Notebook
# Run this entire code in one Jupyter cell.
# Requires: ipywidgets

import random
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# -----------------------------
# State
# -----------------------------
original_heights = [62, 56, 67, 51, 64, 69, 57, 63]  # total inches
heights = original_heights[:]

j = 0
end = len(heights) - 1
pass_number = 1
finished = False
swapped_this_pass = False

# -----------------------------
# UI
# -----------------------------
title = widgets.HTML(
    """
    <h3 style="margin-bottom:4px">Bubble Sort — Height Order</h3>
    <div>Compare two students standing next to each other.
    If the left student is taller, they swap places.</div>
    """
)

status = widgets.HTML()
explanation = widgets.HTML()

next_btn = widgets.Button(description="Next Comparison", button_style="primary")
pass_btn = widgets.Button(description="Finish This Pass")
shuffle_btn = widgets.Button(description="Shuffle")
reset_btn = widgets.Button(description="Reset")

output = widgets.Output()

def format_height(total_inches):
    feet = total_inches // 12
    inches = total_inches % 12
    return f"{feet}' {inches}\""

def display_height(total_inches):
    cm = round(total_inches * 2.54)
    return f"{format_height(total_inches)} ({cm} cm)"

def person_html(height_cm, idx, role="", sorted_pos=False):
    """
    Draw a simple person using only HTML/CSS.
    Visual height is scaled from the student's height in cm.
    """
    # Scale visual body height into a comfortable notebook range
    visual_h = 110 + (height_cm - 50) * 5.0

    if sorted_pos:
        border = "#6aa66a"
        bg = "#e7f7e7"
        role_label = "sorted"
    elif role == "left":
        border = "#d6a600"
        bg = "#fff4c7"
        role_label = "left"
    elif role == "right":
        border = "#4a90e2"
        bg = "#e6f1ff"
        role_label = "right"
    else:
        border = "#aaaaaa"
        bg = "#f7f7f7"
        role_label = ""

    body_h = max(45, visual_h - 62)
    leg_h = max(22, body_h * 0.34)
    torso_h = body_h - leg_h

    return f"""
    <div style="
        width:105px;
        min-height:265px;
        border:3px solid {border};
        background:{bg};
        border-radius:12px;
        margin:6px;
        padding:8px 6px 6px 6px;
        display:flex;
        flex-direction:column;
        align-items:center;
        justify-content:flex-end;
        box-sizing:border-box;
        font-family:Arial, sans-serif;
    ">
        <div style="
            font-size:17px;
            font-weight:bold;
            margin-bottom:4px;
        ">{display_height(height_cm)}</div>

        <div style="
            height:{visual_h}px;
            width:58px;
            display:flex;
            flex-direction:column;
            align-items:center;
            justify-content:flex-end;
        ">
            <!-- head -->
            <div style="
                width:34px;
                height:34px;
                border-radius:50%;
                background:#f2c7a5;
                border:2px solid #555;
                box-sizing:border-box;
                flex-shrink:0;
            "></div>

            <!-- torso -->
            <div style="
                width:38px;
                height:{torso_h}px;
                min-height:38px;
                background:#7aa7d9;
                border:2px solid #555;
                border-radius:8px 8px 5px 5px;
                box-sizing:border-box;
                flex-shrink:0;
            "></div>

            <!-- legs -->
            <div style="
                width:38px;
                height:{leg_h}px;
                display:flex;
                gap:8px;
                justify-content:center;
                flex-shrink:0;
            ">
                <div style="
                    width:8px;
                    height:100%;
                    background:#555;
                    border-radius:3px;
                "></div>
                <div style="
                    width:8px;
                    height:100%;
                    background:#555;
                    border-radius:3px;
                "></div>
            </div>
        </div>

        <div style="
            margin-top:5px;
            font-size:12px;
            min-height:16px;
            font-weight:bold;
        ">{role_label}</div>

        <div style="
            margin-top:2px;
            font-size:11px;
            color:#666;
        ">index {idx}</div>
    </div>
    """

def render():
    global heights, j, end, pass_number, finished

    cards = []

    for idx, h in enumerate(heights):
        if finished or idx > end:
            cards.append(person_html(h, idx, sorted_pos=True))
        elif idx == j:
            cards.append(person_html(h, idx, role="left"))
        elif idx == j + 1:
            cards.append(person_html(h, idx, role="right"))
        else:
            cards.append(person_html(h, idx))

    if finished:
        status.value = "<b>Finished!</b> Everyone is arranged from shortest to tallest."
        explanation.value = (
            "<div style='margin-top:8px'>"
            "<b>What happened?</b> The tallest unsorted student moved toward the right "
            "one comparison at a time during each pass."
            "</div>"
        )
    else:
        left = heights[j]
        right = heights[j + 1]

        status.value = (
            f"Pass <b>{pass_number}</b>. "
            f"Comparing index <b>{j}</b> and <b>{j + 1}</b>: "
            f"<b>{display_height(left)}</b> and <b>{display_height(right)}</b>."
        )

        if left > right:
            action = (
                f"The left student is taller: "
                f"<b>{display_height(left)} &gt; {display_height(right)}</b> → they will <b>swap</b>."
            )
        else:
            action = (
                f"The left student is not taller: "
                f"<b>{display_height(left)} ≤ {display_height(right)}</b> → <b>no swap</b>."
            )

        explanation.value = f"<div style='margin-top:8px'>{action}</div>"

    legend = """
    <div style="
        display:flex;
        gap:18px;
        flex-wrap:wrap;
        margin:8px 0 2px 0;
        font-family:Arial, sans-serif;
        font-size:13px;
    ">
        <span>🟨 left student</span>
        <span>🟦 right student</span>
        <span>🟩 final position</span>
    </div>
    """

    goal = """
    <div style="
        margin-top:8px;
        padding:8px 10px;
        border-left:4px solid #888;
        background:#fafafa;
        font-family:Arial, sans-serif;
    ">
        <b>Goal:</b> Arrange the students from shortest to tallest.
    </div>
    """

    with output:
        clear_output(wait=True)
        display(HTML(
            goal
            + legend
            + "<div style='display:flex;align-items:flex-end;flex-wrap:wrap;'>"
            + "".join(cards)
            + "</div>"
        ))

def reset_state(new_heights=None):
    global heights, j, end, pass_number, finished, swapped_this_pass

    if new_heights is not None:
        heights = new_heights[:]

    j = 0
    end = len(heights) - 1
    pass_number = 1
    finished = False
    swapped_this_pass = False
    render()

def next_comparison(_=None):
    global heights, j, end, pass_number, finished, swapped_this_pass

    if finished:
        return

    if heights[j] > heights[j + 1]:
        heights[j], heights[j + 1] = heights[j + 1], heights[j]
        swapped_this_pass = True

    j += 1

    if j >= end:
        if not swapped_this_pass:
            finished = True
            render()
            return

        end -= 1
        pass_number += 1
        j = 0
        swapped_this_pass = False

        if end <= 0:
            finished = True

    render()

def finish_pass(_=None):
    global heights, j, end, pass_number, finished, swapped_this_pass

    if finished:
        return

    while j < end:
        if heights[j] > heights[j + 1]:
            heights[j], heights[j + 1] = heights[j + 1], heights[j]
            swapped_this_pass = True
        j += 1

    if not swapped_this_pass:
        finished = True
        render()
        return

    end -= 1
    pass_number += 1
    j = 0
    swapped_this_pass = False

    if end <= 0:
        finished = True

    render()

def shuffle_values(_=None):
    new_heights = original_heights[:]
    random.shuffle(new_heights)
    reset_state(new_heights)

def reset_values(_=None):
    reset_state(original_heights)

next_btn.on_click(next_comparison)
pass_btn.on_click(finish_pass)
shuffle_btn.on_click(shuffle_values)
reset_btn.on_click(reset_values)

controls = widgets.HBox([next_btn, pass_btn, shuffle_btn, reset_btn])

display(title, status, controls, explanation, output)
render()
```

```python
# Bubble Sort Planet-Size Visualizer for Jupyter Notebook
# Run this entire code in one Jupyter cell.
# Requires: ipywidgets

import random
import math
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# -----------------------------
# Planet data
# Approximate equatorial diameters in kilometers
# -----------------------------
# Default order: distance from the Sun (nearest -> farthest)
# The sorting key is still planet diameter.
original_planets = [
    {"name": "Mercury", "diameter": 4879,   "symbol": "⚪"},
    {"name": "Venus",   "diameter": 12104,  "symbol": "🟡"},
    {"name": "Earth",   "diameter": 12756,  "symbol": "🌍"},
    {"name": "Mars",    "diameter": 6792,   "symbol": "🔴"},
    {"name": "Jupiter", "diameter": 142984, "symbol": "🟠"},
    {"name": "Saturn",  "diameter": 120536, "symbol": "🪐"},
    {"name": "Uranus",  "diameter": 51118,  "symbol": "🔷"},
    {"name": "Neptune", "diameter": 49528,  "symbol": "🔵"},
]

planets = [p.copy() for p in original_planets]

j = 0
end = len(planets) - 1
pass_number = 1
finished = False
swapped_this_pass = False

# -----------------------------
# UI
# -----------------------------
title = widgets.HTML(
    """
    <h3 style="margin-bottom:4px">Bubble Sort — Planet Size</h3>
    <div>
        Compare two neighboring planets by diameter.
        If the left planet is larger, swap them.
    </div>
    """
)

status = widgets.HTML()
explanation = widgets.HTML()

next_btn = widgets.Button(description="Next Comparison", button_style="primary")
pass_btn = widgets.Button(description="Finish This Pass")
shuffle_btn = widgets.Button(description="Shuffle")
reset_btn = widgets.Button(description="Reset")

output = widgets.Output()

# -----------------------------
# Helpers
# -----------------------------
def format_km(n):
    return f"{n:,} km"

def visual_size(diameter):
    """
    Compress the huge real size differences so all 8 planets remain visible.
    The numeric diameter is the real comparison value.
    """
    min_d = 4879
    max_d = 142984

    # Square-root scaling makes small planets visible while preserving order.
    t = (
        math.sqrt(diameter) - math.sqrt(min_d)
    ) / (
        math.sqrt(max_d) - math.sqrt(min_d)
    )

    return 48 + t * 92

def planet_html(planet, idx, role="", sorted_pos=False):
    size = visual_size(planet["diameter"])

    if sorted_pos:
        border = "#6aa66a"
        bg = "#e7f7e7"
        role_label = "sorted"
    elif role == "left":
        border = "#d6a600"
        bg = "#fff4c7"
        role_label = "left"
    elif role == "right":
        border = "#4a90e2"
        bg = "#e6f1ff"
        role_label = "right"
    else:
        border = "#aaaaaa"
        bg = "#f7f7f7"
        role_label = ""

    return f"""
    <div style="
        width:145px;
        min-height:285px;
        border:3px solid {border};
        background:{bg};
        border-radius:14px;
        margin:6px;
        padding:10px 6px 8px 6px;
        display:flex;
        flex-direction:column;
        align-items:center;
        justify-content:flex-end;
        box-sizing:border-box;
        font-family:Arial, sans-serif;
        text-align:center;
    ">
        <div style="
            font-size:17px;
            font-weight:bold;
            margin-bottom:3px;
        ">{planet["name"]}</div>

        <div style="
            font-size:13px;
            margin-bottom:8px;
        ">Diameter: {format_km(planet["diameter"])}</div>

        <div style="
            width:{size}px;
            height:{size}px;
            border-radius:50%;
            display:flex;
            align-items:center;
            justify-content:center;
            font-size:{max(28, size * 0.48)}px;
            line-height:1;
            margin:auto 0 10px 0;
        ">{planet["symbol"]}</div>

        <div style="
            font-size:11px;
            min-height:18px;
            font-weight:bold;
        ">{role_label}</div>

        <div style="
            margin-top:2px;
            font-size:11px;
            color:#666;
        ">index {idx}</div>
    </div>
    """

def render():
    cards = []

    for idx, planet in enumerate(planets):
        if finished or idx > end:
            cards.append(planet_html(planet, idx, sorted_pos=True))
        elif idx == j:
            cards.append(planet_html(planet, idx, role="left"))
        elif idx == j + 1:
            cards.append(planet_html(planet, idx, role="right"))
        else:
            cards.append(planet_html(planet, idx))

    if finished:
        status.value = (
            "<b>Finished!</b> The planets are arranged "
            "from smallest diameter to largest diameter."
        )
        explanation.value = (
            "<div style='margin-top:8px'>"
            "<b>What happened?</b> During each pass, the largest unsorted planet "
            "moved toward the right one comparison at a time."
            "</div>"
        )
    else:
        left = planets[j]
        right = planets[j + 1]

        status.value = (
            f"Pass <b>{pass_number}</b>. "
            f"Comparing index <b>{j}</b> and <b>{j + 1}</b>: "
            f"<b>{left['name']}</b> ({format_km(left['diameter'])}) and "
            f"<b>{right['name']}</b> ({format_km(right['diameter'])})."
        )

        if left["diameter"] > right["diameter"]:
            action = (
                f"{left['name']} is larger than {right['name']} → "
                f"they will <b>swap</b>."
            )
        else:
            action = (
                f"{left['name']} is not larger than {right['name']} → "
                f"<b>no swap</b>."
            )

        explanation.value = f"<div style='margin-top:8px'>{action}</div>"

    legend = """
    <div style="
        display:flex;
        gap:18px;
        flex-wrap:wrap;
        margin:8px 0 2px 0;
        font-family:Arial, sans-serif;
        font-size:13px;
    ">
        <span>🟨 left planet</span>
        <span>🟦 right planet</span>
        <span>🟩 final position</span>
    </div>
    """

    goal = """
    <div style="
        margin-top:8px;
        padding:8px 10px;
        border-left:4px solid #888;
        background:#fafafa;
        font-family:Arial, sans-serif;
    ">
        <b>Starting order:</b> Distance from the Sun — nearest to farthest.<br>
        Mercury → Venus → Earth → Mars → Jupiter → Saturn → Uranus → Neptune
        <br><br>
        <b>Sorting goal:</b> Rearrange the planets from smallest diameter to largest diameter.
        <br>
        <span style="font-size:12px;color:#666;">
            Note: visual planet sizes are compressed so every planet stays visible.
            The algorithm compares the diameter numbers.
        </span>
    </div>
    """

    with output:
        clear_output(wait=True)
        display(HTML(
            goal
            + legend
            + "<div style='display:flex;align-items:flex-end;flex-wrap:wrap;'>"
            + "".join(cards)
            + "</div>"
        ))

def reset_state(new_planets=None):
    global planets, j, end, pass_number, finished, swapped_this_pass

    if new_planets is not None:
        planets = [p.copy() for p in new_planets]

    j = 0
    end = len(planets) - 1
    pass_number = 1
    finished = False
    swapped_this_pass = False
    render()

def next_comparison(_=None):
    global planets, j, end, pass_number, finished, swapped_this_pass

    if finished:
        return

    if planets[j]["diameter"] > planets[j + 1]["diameter"]:
        planets[j], planets[j + 1] = planets[j + 1], planets[j]
        swapped_this_pass = True

    j += 1

    if j >= end:
        if not swapped_this_pass:
            finished = True
            render()
            return

        end -= 1
        pass_number += 1
        j = 0
        swapped_this_pass = False

        if end <= 0:
            finished = True

    render()

def finish_pass(_=None):
    global planets, j, end, pass_number, finished, swapped_this_pass

    if finished:
        return

    while j < end:
        if planets[j]["diameter"] > planets[j + 1]["diameter"]:
            planets[j], planets[j + 1] = planets[j + 1], planets[j]
            swapped_this_pass = True
        j += 1

    if not swapped_this_pass:
        finished = True
        render()
        return

    end -= 1
    pass_number += 1
    j = 0
    swapped_this_pass = False

    if end <= 0:
        finished = True

    render()

def shuffle_values(_=None):
    new_planets = [p.copy() for p in original_planets]
    random.shuffle(new_planets)
    reset_state(new_planets)

def reset_values(_=None):
    reset_state(original_planets)

next_btn.on_click(next_comparison)
pass_btn.on_click(finish_pass)
shuffle_btn.on_click(shuffle_values)
reset_btn.on_click(reset_values)

controls = widgets.HBox([next_btn, pass_btn, shuffle_btn, reset_btn])

display(title, status, controls, explanation, output)
render()
```

```python

# -----------------------------
# State
# -----------------------------
values = [6, 3, 8, 2, 7, 1, 5, 4]
i = 0                  # start of unsorted region
j = 1                  # current comparison index
min_index = 0          # current minimum index
finished = False

# -----------------------------
# UI
# -----------------------------
title = widgets.HTML(
    "<h3 style='margin-bottom:4px'>Selection Sort Visualizer</h3>"
    "<div>Find the smallest value in the unsorted part, then swap it into place.</div>"
)

status = widgets.HTML()

next_btn = widgets.Button(description="Next Comparison", button_style="primary")
pass_btn = widgets.Button(description="Finish This Pass")
shuffle_btn = widgets.Button(description="Shuffle")
reset_btn = widgets.Button(description="Reset")

output = widgets.Output()

def render():
    global values, i, j, min_index, finished

    boxes = []
    for idx, value in enumerate(values):
        if idx < i:
            bg = "#dff5df"       # sorted
            border = "#6aa66a"
            label = "sorted"
        elif not finished and idx == min_index:
            bg = "#fff0b3"       # current minimum
            border = "#d6a600"
            label = "minimum"
        elif not finished and idx == j:
            bg = "#dcecff"       # comparing
            border = "#4a90e2"
            label = "compare"
        else:
            bg = "#f4f4f4"
            border = "#999"
            label = ""

        boxes.append(
            f"""
            <div style="
                width:70px;
                height:70px;
                display:flex;
                flex-direction:column;
                align-items:center;
                justify-content:center;
                border:3px solid {border};
                background:{bg};
                border-radius:10px;
                margin:5px;
                font-family:Arial;
            ">
                <div style="font-size:26px;font-weight:bold">{value}</div>
                <div style="font-size:11px">{label}</div>
            </div>
            """
        )

    if finished:
        message = "<b>Finished!</b> The list is sorted."
    else:
        message = (
            f"Pass starts at index <b>{i}</b>. "
            f"Current minimum = <b>{values[min_index]}</b> "
            f"(index {min_index}). "
            f"Comparing with index <b>{j}</b>."
        )

    status.value = message

    with output:
        clear_output(wait=True)
        display(HTML(
            "<div style='display:flex;flex-wrap:wrap'>" + "".join(boxes) + "</div>"
        ))

def reset_state(new_values=None):
    global values, i, j, min_index, finished
    if new_values is not None:
        values = new_values[:]
    i = 0
    min_index = 0
    j = 1
    finished = False
    render()

def advance_one_comparison(_=None):
    global values, i, j, min_index, finished

    if finished:
        return

    n = len(values)

    if j < n:
        if values[j] < values[min_index]:
            min_index = j
        j += 1

    # if comparison phase for this pass is complete, perform swap
    if j >= n:
        values[i], values[min_index] = values[min_index], values[i]
        i += 1

        if i >= n - 1:
            finished = True
            render()
            return

        min_index = i
        j = i + 1

    render()

def finish_pass(_=None):
    global values, i, j, min_index, finished

    if finished:
        return

    n = len(values)

    while j < n:
        if values[j] < values[min_index]:
            min_index = j
        j += 1

    values[i], values[min_index] = values[min_index], values[i]
    i += 1

    if i >= n - 1:
        finished = True
    else:
        min_index = i
        j = i + 1

    render()

def shuffle_values(_=None):
    nums = list(range(1, 9))
    random.shuffle(nums)
    reset_state(nums)

def reset_values(_=None):
    reset_state([6, 3, 8, 2, 7, 1, 5, 4])

next_btn.on_click(advance_one_comparison)
pass_btn.on_click(finish_pass)
shuffle_btn.on_click(shuffle_values)
reset_btn.on_click(reset_values)

controls = widgets.HBox([next_btn, pass_btn, shuffle_btn, reset_btn])

display(title, status, controls, output)
render()
```

```python
# Selection Sort Height-Order Visualizer for Jupyter Notebook
# US feet/inches + centimeters
# Run this entire code in one Jupyter cell.
# Requires: ipywidgets

import random
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# -----------------------------
# State
# -----------------------------
original_heights = [62, 56, 67, 51, 64, 69, 57, 63]  # total inches
heights = original_heights[:]

i = 0
j = 1
min_index = 0
finished = False

# -----------------------------
# Height formatting
# -----------------------------
def format_height(total_inches):
    feet = total_inches // 12
    inches = total_inches % 12
    return f"{feet}' {inches}\\\""

def display_height(total_inches):
    cm = round(total_inches * 2.54)
    return f"{format_height(total_inches)} ({cm} cm)"

# -----------------------------
# UI
# -----------------------------
title = widgets.HTML(
    """
    <h3 style="margin-bottom:4px">Selection Sort — Height Order</h3>
    <div>Find the shortest student in the unsorted group.
    After checking everyone, move that student to the first unsorted position.</div>
    """
)

status = widgets.HTML()
explanation = widgets.HTML()

next_btn = widgets.Button(description="Next Comparison", button_style="primary")
pass_btn = widgets.Button(description="Finish This Pass")
shuffle_btn = widgets.Button(description="Shuffle")
reset_btn = widgets.Button(description="Reset")

output = widgets.Output()

def person_html(height, idx, role="", sorted_pos=False):
    visual_h = 110 + (height - 50) * 5.0

    if sorted_pos:
        border = "#6aa66a"
        bg = "#e7f7e7"
        role_label = "sorted"
    elif role == "minimum":
        border = "#d6a600"
        bg = "#fff4c7"
        role_label = "shortest so far"
    elif role == "compare":
        border = "#4a90e2"
        bg = "#e6f1ff"
        role_label = "compare"
    else:
        border = "#aaaaaa"
        bg = "#f7f7f7"
        role_label = ""

    body_h = max(45, visual_h - 62)
    leg_h = max(22, body_h * 0.34)
    torso_h = body_h - leg_h

    return f"""
    <div style="
        width:115px;
        min-height:285px;
        border:3px solid {border};
        background:{bg};
        border-radius:12px;
        margin:6px;
        padding:8px 5px 6px 5px;
        display:flex;
        flex-direction:column;
        align-items:center;
        justify-content:flex-end;
        box-sizing:border-box;
        font-family:Arial, sans-serif;
    ">
        <div style="
            font-size:15px;
            font-weight:bold;
            margin-bottom:4px;
            text-align:center;
        ">{display_height(height)}</div>

        <div style="
            height:{visual_h}px;
            width:58px;
            display:flex;
            flex-direction:column;
            align-items:center;
            justify-content:flex-end;
        ">
            <div style="
                width:34px;
                height:34px;
                border-radius:50%;
                background:#f2c7a5;
                border:2px solid #555;
                box-sizing:border-box;
                flex-shrink:0;
            "></div>

            <div style="
                width:38px;
                height:{torso_h}px;
                min-height:38px;
                background:#7aa7d9;
                border:2px solid #555;
                border-radius:8px 8px 5px 5px;
                box-sizing:border-box;
                flex-shrink:0;
            "></div>

            <div style="
                width:38px;
                height:{leg_h}px;
                display:flex;
                gap:8px;
                justify-content:center;
                flex-shrink:0;
            ">
                <div style="width:8px;height:100%;background:#555;border-radius:3px;"></div>
                <div style="width:8px;height:100%;background:#555;border-radius:3px;"></div>
            </div>
        </div>

        <div style="
            margin-top:5px;
            font-size:11px;
            min-height:28px;
            font-weight:bold;
            text-align:center;
        ">{role_label}</div>

        <div style="font-size:11px;color:#666;">index {idx}</div>
    </div>
    """

def render():
    cards = []

    for idx, h in enumerate(heights):
        if finished or idx < i:
            cards.append(person_html(h, idx, sorted_pos=True))
        elif idx == min_index:
            cards.append(person_html(h, idx, role="minimum"))
        elif idx == j:
            cards.append(person_html(h, idx, role="compare"))
        else:
            cards.append(person_html(h, idx))

    if finished:
        status.value = "<b>Finished!</b> Everyone is arranged from shortest to tallest."
        explanation.value = (
            "<div style='margin-top:8px'><b>Selection Sort:</b> "
            "Each pass found the shortest remaining student and placed that student "
            "at the beginning of the unsorted group.</div>"
        )
    else:
        current_min = heights[min_index]
        compare = heights[j]

        status.value = (
            f"Pass starts at index <b>{i}</b>. "
            f"Shortest so far: <b>{display_height(current_min)}</b> "
            f"(index {min_index}). Comparing with "
            f"<b>{display_height(compare)}</b> (index {j})."
        )

        if compare < current_min:
            action = (
                f"{display_height(compare)} is shorter than "
                f"{display_height(current_min)} → it becomes the "
                f"<b>new shortest student</b>."
            )
        else:
            action = (
                f"{display_height(compare)} is not shorter than "
                f"{display_height(current_min)} → keep the current minimum."
            )

        explanation.value = f"<div style='margin-top:8px'>{action}</div>"

    legend = """
    <div style="
        display:flex;
        gap:18px;
        flex-wrap:wrap;
        margin:8px 0 2px 0;
        font-family:Arial, sans-serif;
        font-size:13px;
    ">
        <span>🟨 shortest so far</span>
        <span>🟦 student being compared</span>
        <span>🟩 final position</span>
    </div>
    """

    goal = """
    <div style="
        margin-top:8px;
        padding:8px 10px;
        border-left:4px solid #888;
        background:#fafafa;
        font-family:Arial, sans-serif;
    ">
        <b>Goal:</b> Arrange the students from shortest to tallest.
    </div>
    """

    with output:
        clear_output(wait=True)
        display(HTML(
            goal
            + legend
            + "<div style='display:flex;align-items:flex-end;flex-wrap:wrap;'>"
            + "".join(cards)
            + "</div>"
        ))

def reset_state(new_heights=None):
    global heights, i, j, min_index, finished

    if new_heights is not None:
        heights = new_heights[:]

    i = 0
    min_index = 0
    j = 1
    finished = False
    render()

def next_comparison(_=None):
    global heights, i, j, min_index, finished

    if finished:
        return

    n = len(heights)

    # Check the current student.
    if heights[j] < heights[min_index]:
        min_index = j

    j += 1

    # Everyone in this unsorted section has now been checked.
    if j >= n:
        heights[i], heights[min_index] = heights[min_index], heights[i]
        i += 1

        if i >= n - 1:
            finished = True
            render()
            return

        min_index = i
        j = i + 1

    render()

def finish_pass(_=None):
    global heights, i, j, min_index, finished

    if finished:
        return

    n = len(heights)

    while j < n:
        if heights[j] < heights[min_index]:
            min_index = j
        j += 1

    heights[i], heights[min_index] = heights[min_index], heights[i]
    i += 1

    if i >= n - 1:
        finished = True
    else:
        min_index = i
        j = i + 1

    render()

def shuffle_values(_=None):
    new_heights = original_heights[:]
    random.shuffle(new_heights)
    reset_state(new_heights)

def reset_values(_=None):
    reset_state(original_heights)

next_btn.on_click(next_comparison)
pass_btn.on_click(finish_pass)
shuffle_btn.on_click(shuffle_values)
reset_btn.on_click(reset_values)

controls = widgets.HBox([next_btn, pass_btn, shuffle_btn, reset_btn])

display(title, status, controls, explanation, output)
render()
```

```python
# Selection Sort Planet-Size Visualizer for Jupyter Notebook
# Default order: nearest to farthest from the Sun
# Sorting goal: smallest to largest planet diameter
# Requires: ipywidgets

import random
import math
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# -----------------------------
# Planet data
# Approximate equatorial diameters in kilometers
# Default order = distance from Sun
# -----------------------------
original_planets = [
    {"name": "Mercury", "diameter": 4879,   "symbol": "⚪"},
    {"name": "Venus",   "diameter": 12104,  "symbol": "🟡"},
    {"name": "Earth",   "diameter": 12756,  "symbol": "🌍"},
    {"name": "Mars",    "diameter": 6792,   "symbol": "🔴"},
    {"name": "Jupiter", "diameter": 142984, "symbol": "🟠"},
    {"name": "Saturn",  "diameter": 120536, "symbol": "🪐"},
    {"name": "Uranus",  "diameter": 51118,  "symbol": "🔷"},
    {"name": "Neptune", "diameter": 49528,  "symbol": "🔵"},
]

planets = [p.copy() for p in original_planets]

i = 0
j = 1
min_index = 0
finished = False

# -----------------------------
# UI
# -----------------------------
title = widgets.HTML(
    """
    <h3 style="margin-bottom:4px">Selection Sort — Planet Size</h3>
    <div>
        Find the smallest planet in the unsorted group.
        After checking all remaining planets, swap it into the first unsorted position.
    </div>
    """
)

status = widgets.HTML()
explanation = widgets.HTML()

next_btn = widgets.Button(description="Next Comparison", button_style="primary")
pass_btn = widgets.Button(description="Finish This Pass")
shuffle_btn = widgets.Button(description="Shuffle")
reset_btn = widgets.Button(description="Reset")

output = widgets.Output()

def format_km(n):
    return f"{n:,} km"

def visual_size(diameter):
    # Compressed visual scale so Mercury remains visible beside Jupiter.
    min_d = 4879
    max_d = 142984
    t = (
        math.sqrt(diameter) - math.sqrt(min_d)
    ) / (
        math.sqrt(max_d) - math.sqrt(min_d)
    )
    return 48 + t * 92

def planet_html(planet, idx, role="", sorted_pos=False):
    size = visual_size(planet["diameter"])

    if sorted_pos:
        border = "#6aa66a"
        bg = "#e7f7e7"
        role_label = "sorted"
    elif role == "minimum":
        border = "#d6a600"
        bg = "#fff4c7"
        role_label = "smallest so far"
    elif role == "compare":
        border = "#4a90e2"
        bg = "#e6f1ff"
        role_label = "compare"
    else:
        border = "#aaaaaa"
        bg = "#f7f7f7"
        role_label = ""

    return f"""
    <div style="
        width:145px;
        min-height:290px;
        border:3px solid {border};
        background:{bg};
        border-radius:14px;
        margin:6px;
        padding:10px 6px 8px 6px;
        display:flex;
        flex-direction:column;
        align-items:center;
        justify-content:flex-end;
        box-sizing:border-box;
        font-family:Arial, sans-serif;
        text-align:center;
    ">
        <div style="font-size:17px;font-weight:bold;margin-bottom:3px;">
            {planet["name"]}
        </div>

        <div style="font-size:13px;margin-bottom:8px;">
            Diameter: {format_km(planet["diameter"])}
        </div>

        <div style="
            width:{size}px;
            height:{size}px;
            border-radius:50%;
            display:flex;
            align-items:center;
            justify-content:center;
            font-size:{max(28, size * 0.48)}px;
            line-height:1;
            margin:auto 0 10px 0;
        ">{planet["symbol"]}</div>

        <div style="
            font-size:11px;
            min-height:30px;
            font-weight:bold;
        ">{role_label}</div>

        <div style="margin-top:2px;font-size:11px;color:#666;">
            index {idx}
        </div>
    </div>
    """

def render():
    cards = []

    for idx, planet in enumerate(planets):
        if finished or idx < i:
            cards.append(planet_html(planet, idx, sorted_pos=True))
        elif idx == min_index:
            cards.append(planet_html(planet, idx, role="minimum"))
        elif idx == j:
            cards.append(planet_html(planet, idx, role="compare"))
        else:
            cards.append(planet_html(planet, idx))

    if finished:
        status.value = (
            "<b>Finished!</b> The planets are arranged "
            "from smallest diameter to largest diameter."
        )
        explanation.value = (
            "<div style='margin-top:8px'>"
            "<b>Selection Sort:</b> Each pass searched all remaining planets, "
            "remembered the smallest one, and then made one swap."
            "</div>"
        )
    else:
        current_min = planets[min_index]
        current = planets[j]

        status.value = (
            f"Pass starts at index <b>{i}</b>. "
            f"Smallest so far: <b>{current_min['name']}</b> "
            f"({format_km(current_min['diameter'])}). "
            f"Comparing with <b>{current['name']}</b> "
            f"({format_km(current['diameter'])})."
        )

        if current["diameter"] < current_min["diameter"]:
            action = (
                f"{current['name']} is smaller than {current_min['name']} → "
                f"<b>{current['name']} becomes the new minimum.</b>"
            )
        else:
            action = (
                f"{current['name']} is not smaller than {current_min['name']} → "
                f"keep <b>{current_min['name']}</b> as the minimum."
            )

        explanation.value = f"<div style='margin-top:8px'>{action}</div>"

    legend = """
    <div style="
        display:flex;
        gap:18px;
        flex-wrap:wrap;
        margin:8px 0 2px 0;
        font-family:Arial, sans-serif;
        font-size:13px;
    ">
        <span>🟨 smallest so far</span>
        <span>🟦 planet being compared</span>
        <span>🟩 final position</span>
    </div>
    """

    goal = """
    <div style="
        margin-top:8px;
        padding:8px 10px;
        border-left:4px solid #888;
        background:#fafafa;
        font-family:Arial, sans-serif;
    ">
        <b>Starting order:</b> Distance from the Sun — nearest to farthest.<br>
        Mercury → Venus → Earth → Mars → Jupiter → Saturn → Uranus → Neptune
        <br><br>
        <b>Sorting goal:</b> Rearrange the planets from smallest diameter to largest diameter.
        <br>
        <span style="font-size:12px;color:#666;">
            Visual planet sizes are compressed so every planet remains visible.
            Selection Sort compares the diameter numbers.
        </span>
    </div>
    """

    with output:
        clear_output(wait=True)
        display(HTML(
            goal
            + legend
            + "<div style='display:flex;align-items:flex-end;flex-wrap:wrap;'>"
            + "".join(cards)
            + "</div>"
        ))

def reset_state(new_planets=None):
    global planets, i, j, min_index, finished

    if new_planets is not None:
        planets = [p.copy() for p in new_planets]

    i = 0
    min_index = 0
    j = 1
    finished = False
    render()

def next_comparison(_=None):
    global planets, i, j, min_index, finished

    if finished:
        return

    n = len(planets)

    # Remember the smallest planet; do NOT swap yet.
    if planets[j]["diameter"] < planets[min_index]["diameter"]:
        min_index = j

    j += 1

    # After checking the whole unsorted section, make one swap.
    if j >= n:
        planets[i], planets[min_index] = planets[min_index], planets[i]
        i += 1

        if i >= n - 1:
            finished = True
            render()
            return

        min_index = i
        j = i + 1

    render()

def finish_pass(_=None):
    global planets, i, j, min_index, finished

    if finished:
        return

    n = len(planets)

    while j < n:
        if planets[j]["diameter"] < planets[min_index]["diameter"]:
            min_index = j
        j += 1

    planets[i], planets[min_index] = planets[min_index], planets[i]
    i += 1

    if i >= n - 1:
        finished = True
    else:
        min_index = i
        j = i + 1

    render()

def shuffle_values(_=None):
    new_planets = [p.copy() for p in original_planets]
    random.shuffle(new_planets)
    reset_state(new_planets)

def reset_values(_=None):
    reset_state(original_planets)

next_btn.on_click(next_comparison)
pass_btn.on_click(finish_pass)
shuffle_btn.on_click(shuffle_values)
reset_btn.on_click(reset_values)

controls = widgets.HBox([next_btn, pass_btn, shuffle_btn, reset_btn])

display(title, status, controls, explanation, output)
render()
```
